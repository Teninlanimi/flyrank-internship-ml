## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

Baseline Rule

Pages that have not been updated for a long time and still receive a high number of impressions should be prioritized for content refresh. These pages are already visible in search results, so updating them may help maintain or improve their performance.

Reason Code

STALE_VISIBLE – The page has not been updated recently but continues to receive strong search visibility.

Action Label

Refresh

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
from pathlib import Path
import pandas as pd

def find_repo_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "data" / "raw" / "content_refresh_anonymized.csv").exists():
            return candidate
    raise FileNotFoundError("Could not find the repo root containing data/raw/content_refresh_anonymized.csv")

repo_root = find_repo_root()
data_path = repo_root / "data" / "raw" / "content_refresh_anonymized.csv"

df = pd.read_csv(data_path)
print(df.shape)
```

```python
import numpy as np

# Rule signals
stale = (df["days_since_last_update"] >= 180)
visible = (df["impressions_90d"] >= 500)

# Baseline score
df["baseline_score"] = np.where(
    stale & visible,
    df["impressions_90d"],
    0
)

# Reason code
df["reason_code"] = np.where(
    stale & visible,
    "STALE_VISIBLE",
    "MONITOR"
)

# Action
df["action"] = np.where(
    stale & visible,
    "Refresh",
    "Monitor"
)

# Rank pages
queue = df.sort_values("baseline_score", ascending=False)

# Save output
queue.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

queue.head(10)
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

Weak picks

My baseline rule prioritizes pages that are stale and still have high visibility. While this is a simple and explainable rule, some recommendations may be incorrect.

Some pages may be selected simply because they have high impressions and have not been updated recently, even though the content is still accurate, evergreen, and does not require a refresh. In addition, some pages may have high visibility due to seasonal search demand rather than outdated content.

I also observed that one of the selected pages (content_bdbec75c1148) had a stable trend rather than a declining one. This suggests that the baseline rule can recommend pages that may not require immediate action because it only considers staleness and visibility, not overall performance trends.

The baseline may also miss pages with low click-through rates or poor engagement because those signals are not included in the rule.

Leakage check

My baseline score only uses observable signals that would have been available before making the recommendation, specifically days_since_last_update and impressions_90d. I did not use product flags or label-derived variables such as trend_direction or trend_pct, since they represent the outcome I am trying to predict and would introduce data leakage. Therefore, the baseline rule does not rely on future information.
